#### <h1>Detect Outliers</h1>

```python
# box plot to visualize monthly income

plt.figure(figsize=(8, 4))

sns.boxplot(x=df['MonthlyIncome'])

plt.title('Boxplot of Monthly Income')
plt.xlabel('Monthly Income')
plt.show()
```

```python
# the IQR method For MonthlyIncome

Q1 = df['MonthlyIncome'].quantile(0.25)
Q3 = df['MonthlyIncome'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
# To check how many potential outliers we have

outliers = df[
    (df['MonthlyIncome'] < lower_bound) |
    (df['MonthlyIncome'] > upper_bound)
]

print("Number of potential outliers:", len(outliers))

print("Maximum MonthlyIncome:", df['MonthlyIncome'].max())
```

```python
# 

df[df['MonthlyIncome'] > upper_bound][
    ['MonthlyIncome', 'JobLevel', 'JobRole', 'TotalWorkingYears', 'YearsAtCompany']
].sort_values('MonthlyIncome', ascending=False).head(10)
```

```python
# screening summary across the important continuous variables

numerical_cols = [
    'Age',
    'DistanceFromHome',
    'TotalWorkingYears',
    'YearsAtCompany',
    'YearsInCurrentRole',
    'YearsSinceLastPromotion',
    'YearsWithCurrManager'
]

for col in numerical_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    count = ((df[col] < lower) | (df[col] > upper)).sum()

    print(f"{col}: {count} potential outliers")
```

#### <h1>Feature Engineering</h1>

```python
# TotalWorkingYears vs YearsAtCompany

#We already have both variables, so we can derive YearsBeforeCompany = TotalWorkingYears − YearsAtCompany

#This represents approximately how many years the employee had worked before joining the current company.

# creating column YearsBeforeCompany

df['YearsBeforeCompany'] = (
    df['TotalWorkingYears'] - df['YearsAtCompany']
)


df[['TotalWorkingYears', 'YearsAtCompany', 'YearsBeforeCompany']].head()
```
